```python
import pandas as pd
df=pd.read_csv(r"C:\Users\rudra\OneDrive\Desktop\sales_dataset.csv")
df
```

```python
df.isnull().sum()
```

```python
df["Region"]=df["Region"].fillna(df["Region"].mode()[0])
df
```

```python
df.isnull().sum()
```

```python
df["Product"]=df["Product"].fillna(df["Product"].mode()[0])
df
```

```python
df.isnull().sum()
```

```python
df['UnitPrice']=df["UnitPrice"].replace("unknown","4")
df
```

```python
df['UnitPrice']=pd.to_numeric(df['UnitPrice'])
df
```

```python
df["UnitPrice"]=df["UnitPrice"].fillna(df["UnitPrice"].mode()[0])
df
```

```python
df.isnull().sum()
```

```python
df["Quantity"]=df["Quantity"].fillna(df["Quantity"].mode()[0])
df
```

```python
df['Quantity']=df["Quantity"].replace("five","5")
df
```

```python
df['Quantity']=pd.to_numeric(df['Quantity'])
df
```

```python
df.dtypes
```

```python
df["Quantity"]=df["Quantity"].fillna(df["Quantity"].mode()[0])
df
```

```python
df.isnull().sum()
```

```python
data=df.groupby("Region")["TotalAmount"].sum().reset_index()
data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.barplot(x="Region",y="TotalAmount",data=data,hue='Region')
plt.show()
```

```python
data=df.groupby("Product")["TotalAmount"].sum().reset_index()
data
```

```python
sns.barplot(x="Product",y="TotalAmount",data=data,hue="Product")
plt.show
```

```python
data=df.groupby("SalesPerson")["Quantity"].sum().reset_index()
data
```

```python
sns.barplot(x="SalesPerson",y="Quantity",data=data,hue='SalesPerson')
plt.show()
```
